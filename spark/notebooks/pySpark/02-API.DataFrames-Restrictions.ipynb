# <b><div style='padding:15px;background-color:#d8dcd6;color:#030aa7;font-size:120%;text-align: left'>Restrictions</div></b>

Une Transformation qui permet de limiter le nombre d’enregistrements du DataFrame résultante est une restriction.

```python
from pyspark.sql.functions import *

meteoDataFrame  = spark.read.format('csv')\
    .option('sep',';')\
    .option('header','true')\
    .option('nullValue','mq')\
    .option('inferSchema', 'true')\
    .load('/opt/spark/donnees/meteo/')  #   .cache()


from pyspark.sql.types     import StructType, \
     StructField, FloatType, \
     IntegerType, StringType

schema = StructType([
        StructField('Id'           , StringType() , True),
        StructField('ville'        , StringType() , True),
        StructField('latitude'     , FloatType() , True),
        StructField('longitude'    , FloatType() , True),
        StructField('altitude'     , IntegerType() , True)])

villes  = spark.read.format('csv')   \
      .option('sep',';')                \
      .option('mergeSchema', 'true')    \
      .option('header','true')          \
      .schema(schema)                   \
      .load('/opt/spark/donnees/postesSynop.csv') #      .cache()

meteoDataFrame.count()
```

## <b><div style='padding:15px;background-color:#d8dcd6;color:#030aa7;font-size:120%;text-align: left'>limit</div></b>

```python
meteoDataFrame.selectExpr(
        'numer_sta','t  - 273.15 as temperature'
        ).limit(5).show()
```

## <b><div style='padding:15px;background-color:#d8dcd6;color:#030aa7;font-size:120%;text-align: left'>distinct</div></b>

```python
meteoDataFrame.select('numer_sta').distinct().count()
```

```python
meteoDataFrame.select('numer_sta').distinct().show(3)
```

```python
meteoDataFrame.distinct().count()
```

## <b><div style='padding:15px;background-color:#d8dcd6;color:#030aa7;font-size:120%;text-align: left'>where</div></b>

```python
villes.where('latitude > 49').show()
```

```python
villes.where('latitude > 49 and altitude > 90').show()
```

```python
villes.where('latitude > 49 or altitude > 800').show()
```

```python
meteoDataFrame.count(), meteoDataFrame.sample(1/100).count(), meteoDataFrame.sample(True,1/100).count()
```

## <b><div style='padding:15px;background-color:#d8dcd6;color:#030aa7;font-size:120%;text-align: left'>sample</div></b>

```python
villes.sample(True,2/100,0).show()
```

## <b><div style='padding:15px;background-color:#d8dcd6;color:#030aa7;font-size:120%;text-align: left'>filter</div></b>

```python
villes.filter("Id == '07168' or Id == '07280'").show()
```

```python
villes.filter("Id = '07168' or Id = '07280'").show()
```
